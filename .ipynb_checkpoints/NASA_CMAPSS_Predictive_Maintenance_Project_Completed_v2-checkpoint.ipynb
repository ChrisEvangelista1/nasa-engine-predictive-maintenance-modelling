## 1. Business understanding

Unexpected equipment failures can cause downtime, missed production, emergency labor, secondary damage, and lost revenue. Preventive maintenance reduces failure risk, but maintaining equipment too early also wastes useful component life and labor.

This project therefore asks two related questions:

### Regression question
> **How many operational cycles remain before an engine fails?**

The target is **Remaining Useful Life (RUL)**, measured in cycles.

### Classification question
> **Is an engine likely to fail within the next 30 cycles?**

This creates an actionable maintenance alert rather than only a numerical life estimate.

### Why both problems matter

A regression model supports planning and prioritization. A classification model is easier to connect to an operational rule such as "inspect this engine". The final threshold should reflect the relative business cost of an unnecessary inspection versus a missed imminent failure.

## 2. About the NASA C-MAPSS data

NASA describes C-MAPSS as a collection of **multivariate engine run-to-failure time series**. Each row is one engine at one operating cycle. The files contain:

- engine / unit identifier
- cycle number
- 3 operational settings
- 21 sensor measurements

Each engine begins in normal operation and eventually develops a fault. In the **training set**, the trajectory continues until failure. In the **test set**, each engine trajectory stops before failure, and NASA supplies the true number of remaining cycles after that stopping point.

We use **FD001** first because it contains:

- 100 training engine trajectories
- 100 test engine trajectories
- 1 operating condition
- 1 fault mode (high-pressure compressor degradation)

That makes it the cleanest C-MAPSS subset for learning the workflow before progressing to FD002–FD004.

**NASA sources**

- Dataset page: https://data.nasa.gov/dataset/cmapss-jet-engine-simulated-data
- NASA Prognostics Center of Excellence repository: https://www.nasa.gov/intelligent-systems-division/discovery-and-systems-health/pcoe/pcoe-data-set-repository/
- Dataset citation: A. Saxena and K. Goebel (2008), *Turbofan Engine Degradation Simulation Data Set*, NASA Prognostics Data Repository.

## 4. Download the dataset

The next cell downloads the current NASA PCoE archive if FD001 is not already present. It also handles nested ZIP files because repository packaging can change.

If your notebook environment blocks downloads:

1. Download **Turbofan Engine Degradation Simulation** from NASA's PCoE repository.
2. Extract the archive.
3. Put the extracted files somewhere under `data/cmapss/`.
4. Rerun the cell.

The code searches recursively, so the exact subfolder structure does not matter.

```python
DATA_DIR = Path("data/cmapss")
DATA_DIR.mkdir(parents=True, exist_ok=True)

NASA_URL = (
    "https://phm-datasets.s3.amazonaws.com/NASA/"
    "6.+Turbofan+Engine+Degradation+Simulation+Data+Set.zip"
)

REQUIRED_FILES = [
    "train_FD001.txt",
    "test_FD001.txt",
    "RUL_FD001.txt",
]


def find_dataset_file(filename, root=DATA_DIR):
    '''Return the first matching file found recursively, or None.'''
    matches = list(root.rglob(filename))
    return matches[0] if matches else None


def extract_nested_zips(root):
    '''Extract ZIP files found under root until no new ZIPs remain.'''
    extracted = set()

    while True:
        zip_files = [p for p in root.rglob("*.zip") if p not in extracted]
        if not zip_files:
            break

        for zip_path in zip_files:
            target = zip_path.parent / zip_path.stem
            target.mkdir(parents=True, exist_ok=True)
            with zipfile.ZipFile(zip_path, "r") as zf:
                zf.extractall(target)
            extracted.add(zip_path)


def ensure_cmapss_data():
    '''Download/extract C-MAPSS only when required files are missing.'''
    if all(find_dataset_file(name) is not None for name in REQUIRED_FILES):
        print("FD001 files already found. No download needed.")
        return

    archive_path = DATA_DIR / "cmapss_nasa.zip"
    print("Downloading NASA C-MAPSS archive...")
    urllib.request.urlretrieve(NASA_URL, archive_path)

    with zipfile.ZipFile(archive_path, "r") as zf:
        zf.extractall(DATA_DIR / "nasa_archive")

    extract_nested_zips(DATA_DIR)

    missing = [name for name in REQUIRED_FILES if find_dataset_file(name) is None]
    if missing:
        raise FileNotFoundError(
            f"Download finished, but these files were not found: {missing}. "
            "Download/extract the NASA archive manually into data/cmapss/."
        )

    print("FD001 downloaded and extracted successfully.")


ensure_cmapss_data()

for name in REQUIRED_FILES:
    print(name, "->", find_dataset_file(name))
```

## 5. Load and label FD001

NASA's text files do not include column headers, so we assign descriptive names. There are 26 columns total: 2 identifiers, 3 operating settings, and 21 sensor measurements.

```python
setting_cols = [f"setting_{i}" for i in range(1, 4)]
sensor_cols = [f"sensor_{i}" for i in range(1, 22)]
column_names = ["engine_id", "cycle"] + setting_cols + sensor_cols

train_path = find_dataset_file("train_FD001.txt")
test_path = find_dataset_file("test_FD001.txt")
rul_path = find_dataset_file("RUL_FD001.txt")

train = pd.read_csv(
    train_path,
    sep=r"\s+",
    header=None,
    names=column_names,
)

test = pd.read_csv(
    test_path,
    sep=r"\s+",
    header=None,
    names=column_names,
)

test_rul = pd.read_csv(
    rul_path,
    sep=r"\s+",
    header=None,
    names=["rul"],
)

test_rul.insert(0, "engine_id", np.arange(1, len(test_rul) + 1))

print("Training shape:", train.shape)
print("Test shape:    ", test.shape)
print("Test RUL rows: ", test_rul.shape)

display(train.head())
```

## 6. Create the Remaining Useful Life target

For the training trajectories, we know the cycle on which each engine fails. Therefore:

\[
RUL = 	ext{failure cycle} - 	ext{current cycle}
\]

For example, if an engine fails at cycle 200, then its RUL at cycle 170 is 30 cycles.

The final training row for every engine has `rul = 0`.

```python
failure_cycle = (
    train.groupby("engine_id")["cycle"]
    .max()
    .rename("failure_cycle")
)

train = train.merge(failure_cycle, on="engine_id", how="left")
train["rul"] = train["failure_cycle"] - train["cycle"]

# Sanity check: every engine should end at RUL = 0.
assert (train.groupby("engine_id")["rul"].min() == 0).all()

print("RUL created successfully.")
display(train[["engine_id", "cycle", "failure_cycle", "rul"]].head(10))
```

## 9. Validation design: split by engine, not by row

This is one of the most important lessons in the project.

A random row split could place early cycles from **Engine 14** in the training set and later cycles from **the same Engine 14** in validation. That makes the validation set less independent and can inflate performance.

Instead, we use `GroupShuffleSplit` with `engine_id` as the group. Entire engines are assigned to either training or validation.

This answers a more realistic question:

> Can the model generalize to engines it has never seen before?

```python
gss = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=RANDOM_STATE,
)

train_idx, val_idx = next(
    gss.split(train, groups=train["engine_id"])
)

reg_train = train.iloc[train_idx].copy()
reg_val = train.iloc[val_idx].copy()

train_engines = set(reg_train["engine_id"])
val_engines = set(reg_val["engine_id"])

assert train_engines.isdisjoint(val_engines)

print(f"Training engines:   {len(train_engines)}")
print(f"Validation engines: {len(val_engines)}")
print(f"Training rows:      {len(reg_train):,}")
print(f"Validation rows:    {len(reg_val):,}")
```

## 10. Define model features

We start with **current-cycle information only**:

- cycle number
- operational settings
- sensor measurements

Any feature that is constant within the model-training engines is removed. This check is performed **after** the engine split so the validation set does not influence feature selection.

```python
raw_candidate_features = ["cycle"] + setting_cols + sensor_cols

raw_features = [
    col for col in raw_candidate_features
    if reg_train[col].nunique(dropna=False) > 1
]

removed_constant_features = sorted(set(raw_candidate_features) - set(raw_features))

print(f"Raw candidate features: {len(raw_candidate_features)}")
print(f"Features retained:      {len(raw_features)}")
print("Removed constant features:", removed_constant_features)
```

## 11. Regression metrics

We will use several metrics because each answers a different question:

- **MAE** — average absolute error in cycles; easiest to explain to stakeholders
- **RMSE** — penalizes large RUL errors more heavily
- **R²** — proportion of target variance explained by the model

For maintenance, MAE and RMSE are usually more operationally interpretable than R².

```python
def regression_metrics(y_true, y_pred):
    '''Return common RUL regression metrics.'''
    return {
        "mae": mean_absolute_error(y_true, y_pred),
        "rmse": mean_squared_error(y_true, y_pred) ** 0.5,
        "r2": r2_score(y_true, y_pred),
    }


def evaluate_regressor(name, model, X_train, y_train, X_val, y_val):
    '''Fit a regression model and return validation metrics plus the fitted model.'''
    start = time.perf_counter()
    model.fit(X_train, y_train)
    fit_time = time.perf_counter() - start

    pred = np.clip(model.predict(X_val), 0, None)
    metrics = regression_metrics(y_val, pred)
    metrics["fit_time_sec"] = fit_time
    metrics["model"] = name

    return metrics, model, pred
```

## 12. Naive regression baseline

A sophisticated model is only useful if it beats a simple baseline.

Our baseline predicts the **median training RUL for every validation row**. It ignores all sensor information.

```python
y_train = reg_train["rul"]
y_val = reg_val["rul"]

baseline_value = y_train.median()
baseline_pred = np.repeat(baseline_value, len(reg_val))

baseline_result = regression_metrics(y_val, baseline_pred)
baseline_result["fit_time_sec"] = 0.0
baseline_result["model"] = "Median baseline"

pd.DataFrame([baseline_result]).set_index("model")
```

## 13. Baseline regression models

We compare models with different assumptions:

### Linear Regression
A simple interpretable benchmark. Scaling is included because it puts the different sensor ranges on a comparable scale.

### Decision Tree
Learns nonlinear rules but can overfit easily.

### Random Forest
A bagged ensemble of trees. Usually more stable than one decision tree.

### XGBoost
A gradient-boosted tree model that learns trees sequentially to correct previous errors. It is often very strong for tabular predictive-maintenance data.

The preprocessing is intentionally inside a `Pipeline` so imputation/scaling is learned only from the training data.

```python
linear_reg = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", LinearRegression()),
])

decision_tree_reg = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("model", DecisionTreeRegressor(
        max_depth=8,
        min_samples_leaf=20,
        random_state=RANDOM_STATE,
    )),
])

random_forest_reg = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("model", RandomForestRegressor(
        n_estimators=300,
        max_depth=None,
        min_samples_leaf=3,
        max_features="sqrt",
        random_state=RANDOM_STATE,
        n_jobs=-1,
    )),
])

xgb_reg = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("model", XGBRegressor(
        n_estimators=400,
        max_depth=4,
        learning_rate=0.05,
        min_child_weight=3,
        subsample=0.85,
        colsample_bytree=0.85,
        objective="reg:squarederror",
        eval_metric="rmse",
        random_state=RANDOM_STATE,
        n_jobs=-1,
    )),
])

regression_models = {
    "Linear Regression": linear_reg,
    "Decision Tree": decision_tree_reg,
    "Random Forest": random_forest_reg,
    "XGBoost": xgb_reg,
}
```

```python
X_train_raw = reg_train[raw_features]
X_val_raw = reg_val[raw_features]

regression_results = [baseline_result]
fitted_regressors = {}
raw_predictions = {}

for name, model in regression_models.items():
    metrics, fitted_model, pred = evaluate_regressor(
        name,
        model,
        X_train_raw,
        y_train,
        X_val_raw,
        y_val,
    )
    regression_results.append(metrics)
    fitted_regressors[name] = fitted_model
    raw_predictions[name] = pred

regression_results_df = (
    pd.DataFrame(regression_results)
    .set_index("model")
    .sort_values("rmse")
)

display(regression_results_df)
```

### Visualize baseline-model predictions

A good RUL model should place predictions reasonably close to the 45-degree line. Pay special attention to the low-RUL region: overestimating life close to failure can be operationally dangerous.

```python
best_raw_name = regression_results_df.drop(index="Median baseline").index[0]
best_raw_pred = raw_predictions[best_raw_name]

plt.figure(figsize=(7, 7))
plt.scatter(y_val, best_raw_pred, alpha=0.25, s=15)
limit = max(y_val.max(), best_raw_pred.max())
plt.plot([0, limit], [0, limit], linestyle="--", linewidth=1.5)
plt.title(f"Actual vs Predicted RUL — {best_raw_name}")
plt.xlabel("Actual RUL")
plt.ylabel("Predicted RUL")
plt.tight_layout()
plt.show()
```

# Part II — Time-series feature engineering

The raw model only sees the current sensor snapshot. But degradation is fundamentally about **change through time**.

Consider two engines that both have a sensor reading of 100 today:

- Engine A has been stable around 100 for 30 cycles.
- Engine B increased from 80 to 100 over the last 10 cycles.

The current value is identical, but the history is very different.

We therefore add five types of historical features:

1. **Lag features** — previous sensor readings
2. **Rolling mean** — recent average sensor level
3. **Rolling standard deviation** — recent instability/variability
4. **Change features** — difference from the prior cycle
5. **Rolling slope** — direction and speed of recent sensor movement

All calculations are done **within each engine** and use only the current and previous cycles. No future measurements are used.

## 14. Select sensors for slope features

Calculating slopes for every sensor is possible, but it adds computation and many correlated features. We choose a small set of sensors that show the strongest monotonic relationship with **cycle** in the model-training engines.

Notice that this selection uses only the training split and does **not** look at validation engines.

```python
def select_trend_sensors(df, sensors, n=8):
    scores = {}

    for sensor in sensors:
        if df[sensor].nunique() <= 1:
            continue

        corr = df[[sensor, "cycle"]].corr(method="spearman").iloc[0, 1]
        if pd.notna(corr):
            scores[sensor] = abs(corr)

    return (
        pd.Series(scores, name="abs_spearman_with_cycle")
        .sort_values(ascending=False)
        .head(n)
    )


trend_sensor_scores = select_trend_sensors(reg_train, sensor_cols, n=8)
trend_sensors = trend_sensor_scores.index.tolist()

display(trend_sensor_scores.to_frame())
print("Slope sensors:", trend_sensors)
```

## 15. Historical feature-engineering function

This function is reusable for training, validation, and test trajectories.

A few details matter:

- `groupby("engine_id")` keeps histories separate across engines.
- `shift()` creates lag features without looking forward.
- rolling windows restart for each engine.
- early-cycle lags/slope values are naturally missing; the model pipeline will median-impute them.

```python
def add_history_features(
    df,
    sensors,
    lag_periods=(1, 5, 10),
    rolling_windows=(5, 20),
    slope_sensors=None,
    slope_window=20,
):
    '''Create past/current-only time-series features within each engine.'''
    out = df.sort_values(["engine_id", "cycle"]).copy()

    # Lag, rolling, and one-cycle change features.
    for sensor in sensors:
        grouped = out.groupby("engine_id")[sensor]

        for lag in lag_periods:
            out[f"{sensor}_lag_{lag}"] = grouped.shift(lag)

        for window in rolling_windows:
            out[f"{sensor}_roll_mean_{window}"] = grouped.transform(
                lambda s: s.rolling(window=window, min_periods=1).mean()
            )
            out[f"{sensor}_roll_std_{window}"] = grouped.transform(
                lambda s: s.rolling(window=window, min_periods=2).std()
            )

        out[f"{sensor}_change_1"] = grouped.diff(1)

    # Rolling slope on a smaller sensor subset for efficiency.
    if slope_sensors is None:
        slope_sensors = []

    def slope(values):
        x = np.arange(len(values))
        if len(values) < 3 or np.all(values == values[0]):
            return 0.0
        return np.polyfit(x, values, 1)[0]

    for sensor in slope_sensors:
        out[f"{sensor}_slope_{slope_window}"] = (
            out.groupby("engine_id")[sensor]
            .transform(
                lambda s: s.rolling(
                    window=slope_window,
                    min_periods=max(3, slope_window // 2),
                ).apply(slope, raw=True)
            )
        )

    return out
```

```python
reg_train_hist = add_history_features(
    reg_train,
    sensors=sensor_cols,
    slope_sensors=trend_sensors,
)

reg_val_hist = add_history_features(
    reg_val,
    sensors=sensor_cols,
    slope_sensors=trend_sensors,
)

# Candidate model columns exclude IDs and targets.
exclude_cols = {"engine_id", "rul", "failure_cycle"}
history_candidate_features = [
    col for col in reg_train_hist.columns
    if col not in exclude_cols
]

# Remove features constant in the model-training engines.
history_features = [
    col for col in history_candidate_features
    if reg_train_hist[col].nunique(dropna=False) > 1
]

print(f"Raw features:        {len(raw_features)}")
print(f"Engineered features: {len(history_features)}")
display(reg_train_hist[history_features[:15]].head())
```

## 16. Test whether historical features improve RUL prediction

This is an important experimental question:

> **Does adding recent sensor history improve prediction compared with using only the current sensor snapshot?**

We will compare Random Forest and XGBoost on the engineered feature set against their raw-feature counterparts.

```python
X_train_hist = reg_train_hist[history_features]
X_val_hist = reg_val_hist[history_features]

y_train_hist = reg_train_hist["rul"]
y_val_hist = reg_val_hist["rul"]

history_models = {
    "Random Forest + history": clone(random_forest_reg),
    "XGBoost + history": clone(xgb_reg),
}

history_results = []
fitted_history_models = {}
history_predictions = {}

for name, model in history_models.items():
    metrics, fitted_model, pred = evaluate_regressor(
        name,
        model,
        X_train_hist,
        y_train_hist,
        X_val_hist,
        y_val_hist,
    )
    history_results.append(metrics)
    fitted_history_models[name] = fitted_model
    history_predictions[name] = pred

history_results_df = pd.DataFrame(history_results).set_index("model")

comparison_df = pd.concat([
    regression_results_df.loc[["Random Forest", "XGBoost"]],
    history_results_df,
]).sort_values("rmse")

display(comparison_df)
```

```python
# Quantify the improvement from raw XGBoost to historical XGBoost.
raw_rmse = regression_results_df.loc["XGBoost", "rmse"]
hist_rmse = history_results_df.loc["XGBoost + history", "rmse"]

rmse_improvement_pct = (raw_rmse - hist_rmse) / raw_rmse * 100

print(f"Raw XGBoost RMSE:        {raw_rmse:.2f} cycles")
print(f"History XGBoost RMSE:    {hist_rmse:.2f} cycles")
print(f"Relative RMSE change:    {rmse_improvement_pct:.1f}%")
```

## 17. Error analysis by proximity to failure

One aggregate RMSE can hide where the model struggles. Maintenance decisions are especially sensitive near failure, so we break validation errors into RUL bands.

```python
best_hist_pred = history_predictions["XGBoost + history"]

error_analysis = reg_val_hist[["engine_id", "cycle", "rul"]].copy()
error_analysis["predicted_rul"] = best_hist_pred
error_analysis["absolute_error"] = np.abs(
    error_analysis["rul"] - error_analysis["predicted_rul"]
)

error_analysis["rul_band"] = pd.cut(
    error_analysis["rul"],
    bins=[-1, 30, 60, 100, np.inf],
    labels=["0–30", "31–60", "61–100", "100+"],
)

band_results = (
    error_analysis.groupby("rul_band", observed=False)
    .agg(
        rows=("rul", "size"),
        actual_rul_mean=("rul", "mean"),
        prediction_mean=("predicted_rul", "mean"),
        mae=("absolute_error", "mean"),
    )
)

display(band_results)
```

# Part III — Hyperparameter tuning with grouped cross-validation

A single validation split is useful for learning and final decision-making, but hyperparameter search is more reliable when it evaluates multiple independent groups.

We use **GroupKFold**, again grouping by `engine_id`. That means an engine can never appear in both the training and validation portion of the same fold.

### Why RandomizedSearchCV instead of GridSearchCV?

XGBoost has many interacting hyperparameters. A full grid can become unnecessarily expensive. Randomized search lets us explore more dimensions with a controlled number of model fits.

> On a slower laptop, reduce `N_ITER` from 15 to 5–10 while learning the workflow.

```python
RUN_TUNING = True
N_ITER = 15

xgb_tuning_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("model", XGBRegressor(
        objective="reg:squarederror",
        eval_metric="rmse",
        random_state=RANDOM_STATE,
        n_jobs=1,  # CV itself is parallelized below.
    )),
])

param_distributions = {
    "model__n_estimators": [250, 400, 600],
    "model__max_depth": [2, 3, 4, 5],
    "model__learning_rate": [0.02, 0.05, 0.10],
    "model__min_child_weight": [1, 3, 5],
    "model__subsample": [0.70, 0.85, 1.00],
    "model__colsample_bytree": [0.70, 0.85, 1.00],
    "model__reg_alpha": [0.0, 0.01, 0.10],
    "model__reg_lambda": [1.0, 5.0, 10.0],
}

group_cv = GroupKFold(n_splits=5)

if RUN_TUNING:
    reg_search = RandomizedSearchCV(
        estimator=xgb_tuning_pipeline,
        param_distributions=param_distributions,
        n_iter=N_ITER,
        scoring="neg_root_mean_squared_error",
        cv=group_cv,
        random_state=RANDOM_STATE,
        n_jobs=-1,
        verbose=1,
        return_train_score=True,
    )

    reg_search.fit(
        X_train_hist,
        y_train_hist,
        groups=reg_train_hist["engine_id"],
    )

    tuned_reg_model = reg_search.best_estimator_

    print("Best grouped-CV RMSE:", -reg_search.best_score_)
    print("Best parameters:")
    display(pd.Series(reg_search.best_params_))
else:
    tuned_reg_model = fitted_history_models["XGBoost + history"]
    print("Tuning skipped; using baseline historical XGBoost.")
```

```python
# Evaluate the selected/tuned model once on the untouched validation engines.
# The validation engines were not used inside RandomizedSearchCV.

tuned_val_pred = np.clip(tuned_reg_model.predict(X_val_hist), 0, None)
tuned_val_metrics = regression_metrics(y_val_hist, tuned_val_pred)

print("Validation performance of selected regression model:")
display(pd.Series(tuned_val_metrics).to_frame("value"))
```

### Interpret SHAP carefully

A feature being important does **not** mean it causes failure. It means the model relies on that feature to make predictions in this dataset.

# Part VI — Reframe the problem as maintenance classification

A regression output such as `RUL = 27` is useful, but many maintenance processes ultimately require a binary action.

We define:

\[
\text{failure\_within\_30} =
\begin{cases}
1 & \text{if RUL} \leq 30\\
0 & \text{otherwise}
\end{cases}
\]

This answers:

> **Should this engine be treated as at risk of failure within the next 30 cycles?**

The value 30 is a business assumption, not a universal rule. In a real organization it should be tied to maintenance lead time, inspection capacity, downtime risk, and component criticality.

```python
FAILURE_WINDOW = 30

cls_train = reg_train_hist.copy()
cls_val = reg_val_hist.copy()

cls_train["failure_within_30"] = (cls_train["rul"] <= FAILURE_WINDOW).astype(int)
cls_val["failure_within_30"] = (cls_val["rul"] <= FAILURE_WINDOW).astype(int)

print("Training class distribution:")
display(
    cls_train["failure_within_30"]
    .value_counts()
    .rename("count")
    .to_frame()
    .assign(proportion=lambda d: d["count"] / d["count"].sum())
)
```

## 18. Classification metrics

For an impending-failure alert, accuracy alone is not enough.

- **Recall:** Of engines/cycles actually within the failure window, how many did we flag?
- **Precision:** Of the alerts we generated, how many were truly within the failure window?
- **F1:** Balance between precision and recall.
- **ROC-AUC:** Ranking quality across thresholds.
- **PR-AUC / Average Precision:** Especially useful when the positive class is less common.

In many maintenance settings, **missing an imminent failure is more expensive than issuing an extra inspection**, so recall often deserves extra attention.

```python
def classification_metrics(y_true, prob, threshold=0.50):
    pred = (prob >= threshold).astype(int)

    return {
        "accuracy": accuracy_score(y_true, pred),
        "precision": precision_score(y_true, pred, zero_division=0),
        "recall": recall_score(y_true, pred, zero_division=0),
        "f1": f1_score(y_true, pred, zero_division=0),
        "roc_auc": roc_auc_score(y_true, prob),
        "pr_auc": average_precision_score(y_true, prob),
    }


def evaluate_classifier(name, model, X_train, y_train, X_val, y_val):
    start = time.perf_counter()
    model.fit(X_train, y_train)
    fit_time = time.perf_counter() - start

    prob = model.predict_proba(X_val)[:, 1]
    metrics = classification_metrics(y_val, prob, threshold=0.50)
    metrics["fit_time_sec"] = fit_time
    metrics["model"] = name

    return metrics, model, prob
```

## 19. Classification model comparison

We reuse the engineered historical features.

Because the failure-window class is not evenly distributed, Logistic Regression and Random Forest use balanced class weights. For XGBoost we calculate `scale_pos_weight` from the training data.

```python
X_cls_train = cls_train[history_features]
y_cls_train = cls_train["failure_within_30"]
X_cls_val = cls_val[history_features]
y_cls_val = cls_val["failure_within_30"]

negative_count = (y_cls_train == 0).sum()
positive_count = (y_cls_train == 1).sum()
scale_pos_weight = negative_count / positive_count

logistic_cls = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=RANDOM_STATE,
    )),
])

random_forest_cls = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("model", RandomForestClassifier(
        n_estimators=300,
        min_samples_leaf=3,
        class_weight="balanced",
        random_state=RANDOM_STATE,
        n_jobs=-1,
    )),
])

xgb_cls = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("model", XGBClassifier(
        n_estimators=350,
        max_depth=4,
        learning_rate=0.05,
        min_child_weight=3,
        subsample=0.85,
        colsample_bytree=0.85,
        scale_pos_weight=scale_pos_weight,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=RANDOM_STATE,
        n_jobs=-1,
    )),
])

classification_models = {
    "Logistic Regression": logistic_cls,
    "Random Forest": random_forest_cls,
    "XGBoost": xgb_cls,
}
```

```python
classification_results = []
fitted_classifiers = {}
classification_probabilities = {}

for name, model in classification_models.items():
    metrics, fitted_model, prob = evaluate_classifier(
        name,
        model,
        X_cls_train,
        y_cls_train,
        X_cls_val,
        y_cls_val,
    )
    classification_results.append(metrics)
    fitted_classifiers[name] = fitted_model
    classification_probabilities[name] = prob

classification_results_df = (
    pd.DataFrame(classification_results)
    .set_index("model")
    .sort_values("pr_auc", ascending=False)
)

display(classification_results_df)
```

## 20. Tune the XGBoost failure classifier with grouped cross-validation

We tune the classifier separately from the RUL regressor because the objective and evaluation metric are different.

For the failure-alert problem, the search optimizes **average precision (PR-AUC)** rather than accuracy. `GroupKFold` again keeps complete engines together so no engine appears in both sides of a fold.

> If runtime is a concern, reduce `N_ITER_CLASSIFIER` to 5 while learning the workflow.

```python
RUN_CLASSIFIER_TUNING = True
N_ITER_CLASSIFIER = 10

xgb_cls_tuning_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("model", XGBClassifier(
        scale_pos_weight=scale_pos_weight,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=RANDOM_STATE,
        n_jobs=1,
    )),
])

cls_param_distributions = {
    "model__n_estimators": [250, 350, 500],
    "model__max_depth": [2, 3, 4, 5],
    "model__learning_rate": [0.02, 0.05, 0.10],
    "model__min_child_weight": [1, 3, 5],
    "model__subsample": [0.70, 0.85, 1.00],
    "model__colsample_bytree": [0.70, 0.85, 1.00],
    "model__reg_alpha": [0.0, 0.01, 0.10],
    "model__reg_lambda": [1.0, 5.0, 10.0],
}

if RUN_CLASSIFIER_TUNING:
    cls_search = RandomizedSearchCV(
        estimator=xgb_cls_tuning_pipeline,
        param_distributions=cls_param_distributions,
        n_iter=N_ITER_CLASSIFIER,
        scoring="average_precision",
        cv=GroupKFold(n_splits=5),
        random_state=RANDOM_STATE,
        n_jobs=-1,
        verbose=1,
        return_train_score=True,
    )

    cls_search.fit(
        X_cls_train,
        y_cls_train,
        groups=cls_train["engine_id"],
    )

    tuned_classifier = cls_search.best_estimator_
    tuned_cls_prob = tuned_classifier.predict_proba(X_cls_val)[:, 1]
    tuned_cls_metrics = classification_metrics(
        y_cls_val,
        tuned_cls_prob,
        threshold=0.50,
    )
    tuned_cls_metrics["fit_time_sec"] = np.nan

    fitted_classifiers["Tuned XGBoost"] = tuned_classifier
    classification_probabilities["Tuned XGBoost"] = tuned_cls_prob
    classification_results_df.loc["Tuned XGBoost"] = tuned_cls_metrics
    classification_results_df = classification_results_df.sort_values(
        "pr_auc", ascending=False
    )

    print("Best grouped-CV PR-AUC:", cls_search.best_score_)
    print("Best parameters:")
    display(pd.Series(cls_search.best_params_))
    print("\nUpdated validation comparison:")
    display(classification_results_df)
else:
    print("Classifier tuning skipped.")
```

### Precision-recall curve

The curve shows the tradeoff created by changing the probability threshold. Lowering the threshold usually increases recall but creates more false alarms.

```python
best_classifier_name = classification_results_df.index[0]
best_classifier = fitted_classifiers[best_classifier_name]
best_val_prob = classification_probabilities[best_classifier_name]

precision, recall, thresholds = precision_recall_curve(y_cls_val, best_val_prob)

plt.figure(figsize=(8, 5))
plt.plot(recall, precision)
plt.title(f"Precision–Recall Curve — {best_classifier_name}")
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.tight_layout()
plt.show()

print("Best classifier at default 0.50 threshold:")
display(pd.Series(
    classification_metrics(y_cls_val, best_val_prob, threshold=0.50)
).to_frame("value"))
```

# Part VII — Threshold optimization using maintenance economics

The model produces probabilities; the business chooses the action threshold.

The following example assumes:

- **False positive:** an unnecessary early inspection costs **$5,000**
- **False negative:** an imminent failure is missed and costs **$100,000**

These are **illustrative assumptions**, not NASA cost figures. Replace them with actual maintenance, downtime, safety, and production costs in a real project.

We select the threshold using **validation data only**. The official NASA test labels remain untouched until final evaluation.

```python
COST_FALSE_POSITIVE = 5_000
COST_FALSE_NEGATIVE = 100_000

threshold_grid = np.arange(0.05, 0.96, 0.01)
threshold_rows = []

for threshold in threshold_grid:
    pred = (best_val_prob >= threshold).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_cls_val, pred, labels=[0, 1]).ravel()

    total_cost = (
        fp * COST_FALSE_POSITIVE
        + fn * COST_FALSE_NEGATIVE
    )

    threshold_rows.append({
        "threshold": threshold,
        "tn": tn,
        "fp": fp,
        "fn": fn,
        "tp": tp,
        "precision": precision_score(y_cls_val, pred, zero_division=0),
        "recall": recall_score(y_cls_val, pred, zero_division=0),
        "f1": f1_score(y_cls_val, pred, zero_division=0),
        "estimated_cost": total_cost,
    })

threshold_results = pd.DataFrame(threshold_rows)
best_threshold_row = threshold_results.loc[
    threshold_results["estimated_cost"].idxmin()
]

selected_threshold = float(best_threshold_row["threshold"])

print(f"Selected cost-minimizing threshold: {selected_threshold:.2f}")
display(best_threshold_row.to_frame("value"))
```

```python
plt.figure(figsize=(9, 5))
plt.plot(threshold_results["threshold"], threshold_results["estimated_cost"])
plt.axvline(selected_threshold, linestyle="--", linewidth=1.5)
plt.title("Estimated Maintenance Decision Cost by Threshold")
plt.xlabel("Failure probability threshold")
plt.ylabel("Estimated validation cost ($)")
plt.tight_layout()
plt.show()
```

### Compare the default and cost-optimized thresholds

This is the bridge between model metrics and business decision-making. A lower-cost threshold may intentionally sacrifice some precision to catch more imminent failures.

```python
threshold_comparison = pd.DataFrame({
    "threshold_0.50": classification_metrics(
        y_cls_val, best_val_prob, threshold=0.50
    ),
    "cost_optimized": classification_metrics(
        y_cls_val, best_val_prob, threshold=selected_threshold
    ),
}).T

display(threshold_comparison)
```

We now refit the selected classifier on the full training fleet. The threshold stays fixed at the value selected using validation engines.

NASA supplies the true final-cycle RUL for each test engine, so we can convert those RUL values into the same `<= 30 cycles` classification target.

```python
full_train_cls = full_train_hist.copy()
full_train_cls["failure_within_30"] = (
    full_train_cls["rul"] <= FAILURE_WINDOW
).astype(int)

X_full_cls = full_train_cls[history_features]
y_full_cls = full_train_cls["failure_within_30"]

final_classifier = clone(best_classifier)
final_classifier.fit(X_full_cls, y_full_cls)

final_test_prob = final_classifier.predict_proba(X_test_last)[:, 1]
final_test_true_cls = (test_rul["rul"] <= FAILURE_WINDOW).astype(int).to_numpy()

final_cls_metrics = classification_metrics(
    final_test_true_cls,
    final_test_prob,
    threshold=selected_threshold,
)

print("NASA final-cycle classification performance:")
display(pd.Series(final_cls_metrics).to_frame("value"))

final_test_cls_pred = (final_test_prob >= selected_threshold).astype(int)
print("Confusion matrix [[TN, FP], [FN, TP]]:")
print(confusion_matrix(final_test_true_cls, final_test_cls_pred, labels=[0, 1]))
```

# Part VIII — Bonus: PCA health index

PCA is an **unsupervised dimensionality-reduction** method. It combines correlated sensor measurements into a smaller number of components.

This does not automatically create a perfect "health score," but it can reveal low-dimensional degradation structure and is useful for visualization.

Good practice: fit the scaler and PCA on the model-training engines only, then transform validation engines.

```python
# Use only nonconstant raw sensors from the training split.
pca_sensor_cols = [
    sensor for sensor in sensor_cols
    if reg_train[sensor].nunique() > 1
]

pca_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("pca", PCA(n_components=min(5, len(pca_sensor_cols)), random_state=RANDOM_STATE)),
])

pca_pipeline.fit(reg_train[pca_sensor_cols])
val_components = pca_pipeline.transform(reg_val[pca_sensor_cols])

explained = pca_pipeline.named_steps["pca"].explained_variance_ratio_

pca_summary = pd.DataFrame({
    "component": [f"PC{i+1}" for i in range(len(explained))],
    "explained_variance_ratio": explained,
})
pca_summary["cumulative_variance"] = pca_summary["explained_variance_ratio"].cumsum()

display(pca_summary)
```

```python
pca_val = reg_val[["engine_id", "cycle", "rul"]].copy()
pca_val["PC1"] = val_components[:, 0]

# Visualize PC1 for a few validation engines.
example_val_engines = sorted(pca_val["engine_id"].unique())[:5]

plt.figure(figsize=(10, 5))
for engine_id in example_val_engines:
    temp = pca_val[pca_val["engine_id"] == engine_id]
    plt.plot(temp["cycle"], temp["PC1"], label=f"Engine {engine_id}")

plt.title("First PCA Component Across Engine Lifecycles")
plt.xlabel("Cycle")
plt.ylabel("PC1 score")
plt.legend()
plt.tight_layout()
plt.show()
```

# Part IX — Bonus: anomaly detection

Predictive maintenance can also be approached without directly predicting RUL.

Here we fit an **Isolation Forest** to the first 20 cycles of the model-training engines as a practical approximation of healthy operation. Then we ask whether later observations increasingly look anomalous.

This is best described as **novelty / anomaly detection using an assumed healthy reference period**, not a completely label-free failure model.

```python
HEALTHY_REFERENCE_CYCLES = 20

anomaly_sensor_cols = pca_sensor_cols

healthy_reference = reg_train.loc[
    reg_train["cycle"] <= HEALTHY_REFERENCE_CYCLES,
    anomaly_sensor_cols,
]

anomaly_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", IsolationForest(
        n_estimators=300,
        contamination="auto",
        random_state=RANDOM_STATE,
        n_jobs=-1,
    )),
])

anomaly_pipeline.fit(healthy_reference)

# sklearn's score_samples is higher for more normal observations.
# Multiply by -1 so larger values mean "more anomalous" for interpretation.
val_anomaly_score = -anomaly_pipeline.score_samples(
    reg_val[anomaly_sensor_cols]
)

anomaly_results = reg_val[["engine_id", "cycle", "rul"]].copy()
anomaly_results["anomaly_score"] = val_anomaly_score
anomaly_results["rul_band"] = pd.cut(
    anomaly_results["rul"],
    bins=[-1, 30, 60, 100, np.inf],
    labels=["0–30", "31–60", "61–100", "100+"],
)

anomaly_by_rul = (
    anomaly_results.groupby("rul_band", observed=False)["anomaly_score"]
    .agg(["count", "mean", "median", "std"])
)

display(anomaly_by_rul)
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(
    anomaly_results["rul"],
    anomaly_results["anomaly_score"],
    alpha=0.20,
    s=12,
)
plt.gca().invert_xaxis()
plt.title("Anomaly Score as Failure Approaches")
plt.xlabel("Remaining Useful Life (decreasing toward failure)")
plt.ylabel("Anomaly score — higher means less normal")
plt.tight_layout()
plt.show()
```

# Part X — Lifecycle-dependent sensor importance

A useful advanced question is:

> **Do the most important signals change as an engine approaches failure?**

We reuse SHAP values from the validation sample and compare average absolute SHAP importance across broad RUL bands.

This is exploratory rather than causal, but it can reveal whether the model relies on different signals in early versus late degradation.

```python
# Align the sampled validation rows with their actual RUL values.
shap_sample_meta = reg_val_hist.loc[shap_sample.index, ["rul"]].copy()
shap_sample_meta["rul_band"] = pd.cut(
    shap_sample_meta["rul"],
    bins=[-1, 30, 60, 100, np.inf],
    labels=["0–30", "31–60", "61–100", "100+"],
)

shap_df = pd.DataFrame(
    np.abs(shap_values),
    columns=history_features,
    index=shap_sample.index,
)
shap_df["rul_band"] = shap_sample_meta["rul_band"]

for band in ["100+", "61–100", "31–60", "0–30"]:
    temp = shap_df.loc[shap_df["rul_band"] == band].drop(columns="rul_band")
    if len(temp) == 0:
        continue

    top = temp.mean().sort_values(ascending=False).head(10)
    print(f"\nTop features for RUL band {band}:")
    display(top.rename("mean_abs_shap").to_frame())
```

# Part XI — Save project outputs

A reproducible project should save the model and the key decision settings needed to recreate predictions. The following cell stores:

- final RUL regression pipeline
- final failure-classification pipeline
- engineered feature names
- selected failure threshold

```python
from IPython.display import Markdown, display

# Describe the direction of the historical-feature experiment.
if history_improvement_pct > 1:
    history_statement = (
        f"Adding lag, rolling, change, and slope features **improved XGBoost RMSE by "
        f"{history_improvement_pct:.1f}%**, from {raw_xgb_rmse:.2f} to "
        f"{history_xgb_rmse:.2f} cycles. This supports the idea that degradation "
        f"trajectory contains useful information beyond the current sensor snapshot."
    )
elif history_improvement_pct >= -1:
    history_statement = (
        f"Historical features changed XGBoost RMSE only slightly "
        f"({raw_xgb_rmse:.2f} to {history_xgb_rmse:.2f} cycles; "
        f"{history_improvement_pct:.1f}%). The extra complexity did not create a "
        f"material improvement, so feature selection or alternative history windows "
        f"would be worth testing before keeping the full engineered set."
    )
else:
    history_statement = (
        f"Historical features **reduced performance** in this run: XGBoost RMSE moved "
        f"from {raw_xgb_rmse:.2f} to {history_xgb_rmse:.2f} cycles "
        f"({history_improvement_pct:.1f}%). This suggests that the engineered history "
        f"features added noise, redundancy, or unnecessary complexity and should be "
        f"revisited rather than assumed to be beneficial."
    )

# Describe the maintenance-threshold direction.
if selected_threshold < 0.50:
    threshold_direction = (
        "The selected threshold is below 0.50, which intentionally makes the system "
        "more willing to issue an alert in order to reduce costly missed failures."
    )
elif selected_threshold > 0.50:
    threshold_direction = (
        "The selected threshold is above 0.50, which makes the system more selective "
        "about issuing alerts and places relatively more emphasis on avoiding false alarms."
    )
else:
    threshold_direction = (
        "The cost analysis selected approximately the conventional 0.50 threshold for "
        "the assumptions used in this notebook."
    )

# Describe anomaly behavior without assuming the direction in advance.
if near_failure_anomaly > early_life_anomaly:
    anomaly_statement = (
        f"The mean anomaly score was higher close to failure "
        f"({near_failure_anomaly:.3f} for RUL 0–30 versus {early_life_anomaly:.3f} "
        f"for RUL 100+), providing exploratory evidence that the sensor state becomes "
        f"less similar to the assumed healthy reference period as failure approaches."
    )
else:
    anomaly_statement = (
        f"The anomaly score did not increase close to failure in this run "
        f"({near_failure_anomaly:.3f} for RUL 0–30 versus {early_life_anomaly:.3f} "
        f"for RUL 100+). The simple Isolation Forest approach therefore should not be "
        f"treated as a reliable standalone failure indicator without further work."
    )

shap_feature_text = ", ".join(f"`{feature}`" for feature in top_shap_features)

conclusion_text = f"""
## Conclusions

### 1. The project established a meaningful predictive-maintenance baseline

The median-only baseline produced a validation **MAE of {baseline_mae:.2f} cycles** and **RMSE of {baseline_rmse:.2f} cycles**. The best raw-feature machine-learning model was **{best_raw_model_name}**, with a validation RMSE of **{best_raw_rmse:.2f} cycles**. This comparison is important because a complex model is only useful if it improves on a simple rule that ignores the sensor data.

After model selection, grouped cross-validation, and refitting on the complete training fleet, the final RUL model achieved an official NASA FD001 test **MAE of {final_test_mae:.2f} cycles**, **RMSE of {final_test_rmse:.2f} cycles**, and **R² of {final_test_r2:.3f}** at the final observed cycle of each test engine. These test results are the most important regression numbers in the notebook because they measure generalization to engine trajectories that were not used for model development.

### 2. Sensor history was tested as an explicit hypothesis

{history_statement}

This before/after comparison is more informative than simply engineering a large number of features and assuming they help. It directly tests whether recent **direction, variability, and trajectory** add predictive information.

### 3. Prediction error is not equally important at every point in an engine's life

The highest MAE occurred in the **{worst_rul_band} cycle RUL band**, where MAE was **{worst_band_mae:.2f} cycles**. Within the operationally important **0–30 cycle** window, MAE was **{near_failure_mae:.2f} cycles**.

For maintenance decisions, errors close to failure deserve special attention. Overestimating an engine's remaining life when only a few cycles remain can create more operational risk than making the same numerical error much earlier in the engine lifecycle. Future model selection should therefore consider **where errors occur**, not only the overall RMSE.

### 4. Model explainability identified the signals driving RUL predictions

The five highest global SHAP features in this run were {shap_feature_text}. Their importance indicates that the model relies heavily on these signals when estimating RUL. Engineered lag, rolling, change, or slope features appearing near the top would strengthen the evidence that recent degradation behavior matters.

SHAP should still be interpreted as **model explanation rather than physical causation**. A feature can be highly predictive because it tracks degradation without being the mechanical cause of the failure itself.

### 5. The failure-within-30-cycles model provides a more actionable maintenance output

The strongest validation classifier was **{best_classifier_name}**, with a PR-AUC of **{default_cls_metrics['pr_auc']:.3f}**. At the default 0.50 threshold it achieved **precision {default_cls_metrics['precision']:.3f}** and **recall {default_cls_metrics['recall']:.3f}**.

Using the illustrative maintenance costs of **${COST_FALSE_POSITIVE:,.0f} per false positive** and **${COST_FALSE_NEGATIVE:,.0f} per false negative**, the validation analysis selected a probability threshold of **{selected_threshold:.2f}**. At that threshold, validation precision was **{optimized_cls_metrics['precision']:.3f}** and recall was **{optimized_cls_metrics['recall']:.3f}**. The estimated validation decision cost changed from **${default_estimated_cost:,.0f}** near the 0.50 threshold to **${optimized_estimated_cost:,.0f}**, a reduction of approximately **{estimated_cost_reduction_pct:.1f}%** under these assumptions.

{threshold_direction}

On the official NASA test engines, the final classifier achieved **precision {final_cls_metrics['precision']:.3f}**, **recall {final_cls_metrics['recall']:.3f}**, **F1 {final_cls_metrics['f1']:.3f}**, and **PR-AUC {final_cls_metrics['pr_auc']:.3f}** using the threshold chosen from validation data.

### 6. PCA and anomaly detection provided useful exploratory views, but not replacement models

The first two principal components explained **{pca_two_component_variance:.1%}** of the variance in the selected non-constant sensor measurements. PCA is useful for visualizing lower-dimensional engine behavior, but a principal component should not automatically be interpreted as a physical "health score."

{anomaly_statement}

The anomaly analysis is best treated as exploratory because the "healthy" reference period was approximated using the first {HEALTHY_REFERENCE_CYCLES} cycles rather than a separately labeled healthy population.

---

## Business recommendations

1. **Use RUL predictions for maintenance planning and prioritization.** Rank engines by predicted remaining life so maintenance teams can focus attention on equipment approaching the end of its useful operating window rather than relying only on fixed service intervals.

2. **Use the 30-cycle classifier as a separate near-term alert.** The regression model answers *how long is left?* while the classifier answers *does this engine require near-term attention?* Using both outputs provides more operational context than either model alone.

3. **Choose alert thresholds from business costs and maintenance capacity.** The default 0.50 cutoff should not be treated as a rule. Thresholds should reflect the cost of missed failures, unnecessary inspections, downtime, spare-part availability, labor capacity, and equipment criticality.

4. **Monitor the most influential sensor signals and their trends.** The SHAP results provide a starting point for determining which signals deserve closer engineering review, dashboarding, and sensor-quality monitoring.

5. **Prioritize performance close to failure.** Model monitoring should separately track errors in low-RUL bands. A model with a slightly better overall RMSE may still be less useful if it systematically overestimates life near failure.

6. **Keep a human maintenance decision in the loop.** The model should initially support inspection and scheduling decisions rather than automatically removing equipment from service. Predictions should be combined with engineering knowledge, inspection results, safety requirements, and maintenance history.

---

## Limitations

- **C-MAPSS is simulated data.** Real equipment will contain additional noise, sensor failures, repairs, environmental variation, maintenance interventions, and data-quality problems.
- **FD001 is the simplest C-MAPSS scenario.** It contains one operating condition and one fault mode, so performance should not be assumed to transfer directly to more complex equipment environments.
- **RUL is modeled as a point estimate.** The current workflow does not provide a prediction interval showing how uncertain each RUL estimate is.
- **The 30-cycle failure window is a business assumption.** A real maintenance horizon should be based on lead time, maintenance planning, safety requirements, and component criticality.
- **The maintenance costs are hypothetical.** The selected threshold is only optimal for the example ${COST_FALSE_POSITIVE:,.0f}/${COST_FALSE_NEGATIVE:,.0f} cost assumptions and the observed validation data.
- **SHAP is not causal analysis.** Important features explain model behavior but do not prove which physical mechanisms caused degradation.
- **The anomaly detector uses an assumed healthy period.** The first {HEALTHY_REFERENCE_CYCLES} cycles are a convenient reference, not a verified healthy-state label.
- **Repeated cycles from each engine are correlated.** Grouped splitting protects against engine leakage, but the observations are still longitudinal rather than independent samples.

---

## Recommended next steps

### 1. Test a capped or piecewise RUL target
Early in an engine's life, the exact number of cycles to eventual failure may be difficult to infer from sensors because degradation has not become visible yet. A common follow-up experiment is to cap high RUL values and focus the regression target on the degradation period. Compare the capped-target model against the raw-RUL model using the same engine-level validation design.

### 2. Improve the historical feature set
Test additional time-aware features such as exponentially weighted moving averages, rolling minimum/maximum values, sensor-to-baseline deviations, and different lag/rolling windows. Keep each experiment tied to validation performance so feature engineering does not become uncontrolled feature expansion.

### 3. Add uncertainty estimates
Maintenance planners benefit from knowing whether a prediction is approximately **25 cycles ± 3** or **25 cycles ± 20**. Quantile regression or conformal prediction would extend the project from point predictions to prediction intervals without requiring neural networks.

### 4. Predict multiple maintenance horizons
Instead of only `failure_within_30`, compare alerts such as **15, 30, 60, and 90 cycles**. Different horizons could support different actions—for example, parts ordering, inspection scheduling, and immediate maintenance escalation.

### 5. Calibrate failure probabilities
Check whether predicted probabilities correspond to observed failure frequencies. Probability calibration is especially important if the model output will feed a cost model or maintenance-priority score.

### 6. Progress through FD002, FD003, and FD004
- **FD002:** six operating conditions, one fault mode — learn to normalize or model sensor behavior by operating regime.
- **FD003:** one operating condition, two fault modes — test whether the same features work across different degradation mechanisms.
- **FD004:** six operating conditions and two fault modes — combine both sources of complexity in the strongest traditional-ML extension of this project.

### 7. Explore survival-analysis methods
Survival analysis provides another way to model time-to-failure and naturally handles censored observations. It would be a valuable extension after the standard regression/classification workflow is fully understood.

### 8. Design a deployment-monitoring plan
A production version should monitor missing sensors, sensor-range changes, feature drift, predicted-RUL distributions, alert volume, precision/recall after outcomes become known, and changes in maintenance policy. Model retraining should be triggered by evidence of degraded performance rather than by an arbitrary schedule alone.

---

## Overall conclusion

This project demonstrates more than simply fitting an XGBoost model. It builds a **leakage-aware predictive-maintenance workflow** that connects multivariate sensor data, time-series feature engineering, grouped validation, regression, classification, explainability, anomaly detection, and cost-sensitive decision-making.

The most important practical lesson is that predictive maintenance is not only a question of maximizing a model metric. A useful system must identify deterioration early enough to support action, control the cost of false alarms and missed failures, explain which signals are influencing predictions, and continue to perform on completely unseen equipment. The NASA FD001 workflow provides a strong foundation for those concepts while remaining manageable with traditional machine-learning methods.
"""

display(Markdown(conclusion_text))
```
